```python
import pandas as pd

# Caminho do arquivo
arquivo = '/content/tabela9942 (2).xlsx'

# Carregar o arquivo Excel para inspecionar os nomes das abas (sheets)
excel_file = pd.ExcelFile(arquivo)
print("Abas encontradas no arquivo:", excel_file.sheet_names)

# Substitua "Tabela" pelo nome correto de uma das abas listadas acima
nome_da_aba = excel_file.sheet_names[0]  # Por padrão, vamos tentar ler a primeira aba

# Ler a planilha sem interpretar os cabeçalhos automaticamente
bruto = pd.read_excel(
    arquivo,
    sheet_name=nome_da_aba,
    header=None
)

# Visualizar as primeiras linhas e colunas
display(bruto.head(12))
```

```python

# Identificar os cabeçalhos do IBGE
cabecalho = bruto.iloc[3:6].copy()

# Preencher células mescladas horizontalmente
cabecalho = cabecalho.ffill(axis=1)

# Criar nomes para as colunas de dados
nomes_colunas = []

for i in range(2, bruto.shape[1]):

    partes = [
        str(valor).strip()
        for valor in cabecalho.iloc[:, i]
        if pd.notna(valor)
    ]

    nomes_colunas.append(" | ".join(partes))

# Separar os dados
dados = bruto.iloc[6:].copy()

# Definir nomes das colunas
dados.columns = [
    "municipio",
    "faixa_etaria"
] + nomes_colunas

# Preencher municípios eventualmente mesclados
dados["municipio"] = dados["municipio"].ffill()

# Manter apenas Bom Jesus de Goiás
dados = dados[
    dados["municipio"].astype("string").str.contains(
        "Bom Jesus de Goiás",
        case=False,
        na=False
    )
].copy()

display(dados)
```

```python

# Transformar formato largo em formato longo
df = dados.melt(
    id_vars=["municipio", "faixa_etaria"],
    var_name="classificacao",
    value_name="moradores"
)

# Separar as classificações
partes = df["classificacao"].str.split(
    " | ",
    n=2,
    expand=True,
    regex=False
)

partes.columns = [
    "ano",
    "conexao_internet",
    "cor_raca"
]

# Juntar as informações
df = pd.concat(
    [df.drop(columns="classificacao"), partes],
    axis=1
)

display(df.head(20))
```

```python

# Ajustar textos
colunas_texto = [
    "municipio",
    "faixa_etaria",
    "conexao_internet",
    "cor_raca"
]

for coluna in colunas_texto:
    df[coluna] = df[coluna].astype("string").str.strip()

# Converter ano para número
df["ano"] = pd.to_numeric(
    df["ano"],
    errors="coerce"
).astype("Int64")

# Converter quantidade de moradores
df["moradores"] = pd.to_numeric(
    df["moradores"],
    errors="coerce"
)

# Remover registros sem valores numéricos
df = df.dropna(subset=["moradores"])

# Remover duplicações
df = df.drop_duplicates()

# Organizar as linhas
df = df.sort_values(
    ["ano", "faixa_etaria", "cor_raca"]
).reset_index(drop=True)

display(df)
```

```python

print("Quantidade de linhas e colunas:", df.shape)

print("\nInformações da base:")
df.info()

print("\nValores ausentes:")
print(df.isnull().sum())

print("\nCategorias de conexão:")
print(df["conexao_internet"].unique())

print("\nCategorias de cor ou raça:")
print(df["cor_raca"].unique())

print("\nFaixas etárias:")
print(df["faixa_etaria"].unique())
```

```python
# Colunas nas quais queremos excluir a categoria Total
colunas = [
    "faixa_etaria",
    "conexao_internet",
    "cor_raca"
]

# Remover linhas com Total em qualquer uma dessas colunas
df = df[
    ~df[colunas].apply(
        lambda x: x.str.strip().str.lower().eq("total")
    ).any(axis=1)
].copy()

# Reiniciar a numeração das linhas
df = df.reset_index(drop=True)

display(df)
```

```python

# Salvar base tratada
df.to_csv(
    "/content/tabela9942_limpa.csv",
    index=False,
    encoding="utf-8-sig"
)

# Baixar arquivo
from google.colab import files

files.download("/content/tabela9942_limpa.csv")
```
